```python
#load the data
def load_data(file_name):
    with open(file_name,"r") as f:
        k=json.load(f)
        return k
```

```python
p=load_data("store_data.json")
print(p,type(p))
```

```python
def cleandata(data):
    for users in data:
        print(users)
```

```python
cleandata(p)
```

```python
#clean and structure the data
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    s=set()
    l=[]
    for users in data:
        #clean rating
        q=users["rating"]
        if(type(q)==str):
            users["rating"]=text_to_num[q]
        print(users["rating"])
        #handle missing values
        if(users.get("age")==None):
            users["age"]=None
        #handle dublication
        if users["name"] in s:
            continue
        
        s.add(users["name"])
        l.append(users)
        print(users)
    return l
```

```python
clean_data(p)
```

```python
print(p)
```

```python
k=clean_data(p)
print(k)
```

```python
def avg_rating(data):
    count=len(data)
    sum=0
    for users in data:
        sum+=users["rating"]
    ans=sum/count
    return ans
        
```

```python
def percentage(data):
    count=len(data)
    sum=0
    for users in data:
        if(users["rating"]<3):
            sum+=1
    ans=sum*100/count
    return ans
        
```

```python
def recommendation(data):
    dic={}
    for users in data:
        if(users["rating"]>=4):
            dic[users["name"]]="Apple"
        else:
            dic[users["name"]]="Samsung"
    return dic   
            
```

```python
m=clean_data(p)
print(m)
ans1=avg_rating(m)
ans2=percentage(m)
print(f"the avg rating : {ans1}{"\n"}the percentage of users gave poor rating <3 : {ans2}%")
ans3=recommendation(m)
print(f"user recommendation are {"\n"}{ans3}")
```
